```python
#Q3. You are managing a list of students enrolled in two different courses. You need to perform various set operations to understand the student enrollment. Task: Create two sets: one for students in "Course A" and one for students in "Course B". Find the students who are in "Course A" but not in "Course B" (difference). Find students who are only in one of the two courses (symmetric difference). course_a = {"John", "Alice", "Bob", "David"} course_b = {"Alice", "Eve", "Charlie", "David"}

#Using set methods

#• Find students who are in Course A but not in Course B (difference)

#• Find students who are only in one of the two courses (symmetric difference)

#Expected Output:

#Students in Course A but not in Course B: {'Bob', 'John'}

#Students only in one of the two courses: {'Eve', 'John', 'Bob', 'Charlie'}

course_a = {"John", "Alice", "Bob", "David"}
course_b = {"Alice", "Eve", "Charlie", "David"}

difference = course_a.difference(course_b)
symmetric_difference = course_a.symmetric_difference(course_b)

print("Students in Course A but not in Course B:", difference)
print("Students only in one of the two courses:", symmetric_difference)
```

```python
#Q5. Combine Two Sets
#A = {"Ali", "Sara"}
#B = {"Ahmed", "Zara"}

#Combine both sets into one set.

A = {"Ali", "Sara"}
B = {"Ahmed", "Zara"}

combined = A.union(B)

print("Combined Set:", combined)
```

```python
#Q6. Set Comparison
#A = {1, 2, 3, 4, 5}
#B = {4, 5, 6, 7}

#Find common elements.
#Find elements only in Set A.
#Find elements only in Set B.
    
A = {1, 2, 3, 4, 5}
B = {4, 5, 6, 7}

print("Common elements:", A.intersection(B))
print("Only in Set A:", A.difference(B))
print("Only in Set B:", B.difference(A))
```

```python
#Q7. Given the tuple my_tuple = (1, 2, 3, 4, 5), remove the element 3 from the tuple by first converting it to a list, and then print the resulting tuple.

my_tuple = (1, 2, 3, 4, 5)

my_list = list(my_tuple)
my_list.remove(3)

my_tuple = tuple(my_list)

print(my_tuple)
```
